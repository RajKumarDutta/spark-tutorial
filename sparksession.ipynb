##### Spark Session Builder

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
```

```python
# Create Spark session
spark = SparkSession.builder.appName("netflix recommendation").getOrCreate()
```

```python
df = spark.read.csv("netflix_titles.csv", header=True, inferSchema=True)
df.show(5)
```

```python
df.printSchema()
```

```python
df.count()
```

```python
movies = df.filter(df.type == "Movie")
movies.show(5)
```

```python
df.select("title", "release_year").show(5)
```

```python
df.groupBy("type").count().show()
```

```python
df.na.drop().show(5)
```

```python
from pyspark.sql.functions import col

df = df.withColumn("year_plus_1", col("release_year") + 1)
df.show(5)
```

```python
df.write.csv("output", header=True)
```
